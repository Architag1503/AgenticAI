```python
from langgraph.graph import StateGraph , START , END
from langchain_mistralai import ChatMistralAI
from typing import TypedDict
from dotenv import load_dotenv
from rich import print
```

```python
load_dotenv()

model = ChatMistralAI(model="mistral-small-2506")
```

```python
class BlogState(TypedDict):
    topic: str
    outline: str
    content: str
```

```python
def create_outline(state: BlogState):
    # fetch the topic
    topic = state['topic']

    # generate the outline
    outline = model.invoke(f"generate the outline for a blog on topic: {topic}")

    # update the state
    state['outline'] = outline

    return state
```

```python
def create_blog(state: BlogState):

    # fetch the outline
    topic = state['topic']
    outline = state['outline']

    # generate the blog
    blog = model.invoke(f"Generate the blog on topic: {topic} and using the outline: {outline}")

    # Update the state
    state['content'] = blog

    return state
```

```python
graph = StateGraph(BlogState)

# create the nodes
graph.add_node('create_outline' , create_outline)
graph.add_node('create_blog' , create_blog)

# Add the edges
graph.add_edge(START , 'create_outline')
graph.add_edge('create_outline' , 'create_blog')
graph.add_edge('create_blog' , END)

# compile the graph
workflow = graph.compile()
```

```python
# Execute the workflow
res = workflow.invoke({'topic' : 'LangGraph'})
print(res)
```

```python
print(res['topic'])
print(res['outline'].content)
print(res['content'].content)
```
