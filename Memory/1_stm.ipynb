```python
# Import necessary modules from LangGraph to create a state graph and define states
from langgraph.graph import StateGraph, START, MessagesState
# Import Mistral AI chat model integration from LangChain
from langchain_mistralai import ChatMistralAI
# Import load_dotenv to load environment variables (like API keys) from a .env file
from dotenv import load_dotenv
```

```python
# Load environment variables into the system so that ChatMistralAI can access the API key
load_dotenv()
```

```python
# Initialize the Mistral AI model, specifying the particular model version to use
model = ChatMistralAI(model="mistral-small-2506")
```

```python
# Define a node function for our graph
# This function takes the current state (which includes a list of messages) and passes them to the model
def call_model(state: MessagesState):
    # The model processes the conversation history and generates a response
    response = model.invoke(state["messages"])
    # The response is returned to be appended to the state's message list
    return {"messages": [response]}
```

```python
# Initialize a StateGraph with MessagesState. MessagesState is a built-in state type 
# that automatically appends new messages to the existing list of messages.
builder = StateGraph(MessagesState)
# Add our 'call_model' function as a node in the graph
builder.add_node("call_model", call_model)
# Add an edge from the predefined START node to our 'call_model' node
# This means the graph will always start by executing the 'call_model' node
builder.add_edge(START, "call_model")
```

```python
# Compile the graph into an executable state machine
graph = builder.compile()
graph
```

```python
# Invoke the graph with an initial user message. 
# Without memory, the graph forgets this conversation immediately after it finishes.
graph.invoke({"messages": [{"role": "user", "content": "Hi! My name is Archit."}]})
```

```python
# If we ask the model for the name again, it will not know it,
# because the graph has no memory of the previous invocation.
graph.invoke({"messages": [{"role": "user", "content": "What is my name?"}]})
```

```python
# Let's rebuild the graph to add Short-Term Memory (STM)
builder = StateGraph(MessagesState)
builder.add_node("call_model", call_model)
builder.add_edge(START, "call_model")
```

```python
# Import InMemorySaver, which stores the graph's state (memory) in RAM
from langgraph.checkpoint.memory import InMemorySaver

# Initialize the checkpointer
checkpointer = InMemorySaver()
# Compile the graph again, but this time pass the checkpointer.
# This enables the graph to persist state between invocations.
graph = builder.compile(checkpointer=checkpointer)
```

```python
# Create configurations for two separate conversation threads.
# 'thread_id' acts as a unique identifier for a specific conversation session.
config = {"configurable": {"thread_id": "thread-1"}}
config2 = {"configurable": {"thread_id": "thread-2"}}
```

```python
# Invoke the graph using the first thread config.
# The checkpointer will save this conversation under 'thread-1'.
graph.invoke({"messages": [{"role": "user", "content": "Hi! My name is Archit."}]}, config)
```

```python
# When we ask for the name using a DIFFERENT thread (thread-2),
# the model doesn't know it, because thread-2 has its own isolated memory.
graph.invoke({"messages": [{"role": "user", "content": "What is my name?"}]}, config2)
```

```python
# We can manually retrieve the state (memory) of a specific thread.
# Here we get the state for thread-1 and print all messages in its history.
snap = graph.get_state(config)
vals = snap.values
for m in vals.get("messages", []):
        print("-", type(m).__name__, ":", m.content)
```
