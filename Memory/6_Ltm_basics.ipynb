```python
# InMemoryStore is a simple, non-persistent key-value store used here to demonstrate LTM concepts.
from langgraph.store.memory import InMemoryStore
```

```python
# creating a store
store = InMemoryStore()
```

```python
# creating a namespace
namespace = ("user", "u1")
```

### Creating Memories
Memories in LangGraph are stored within a specific `namespace`. A namespace (e.g., `('user', 'u1')`) logically separates memory for different users or contexts.
- `store.put(namespace, key, value)` is used to save memory items.
- The value is usually a dictionary containing a `data` key with the actual memory content.

```python
# Adding memories
store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data": "User prefers dark mode"})
```

```python
	
# another namespace
namespace2 = ("user", "u2")

# Add memories
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

### Retrieving Memories
- `store.get(namespace, key)` fetches a specific memory item by its unique key within the given namespace.

```python
# store.get(namespace, key)
store.get(namespace2, "2")
```

### Retrieving all Memories
- `store.search(namespace)` retrieves all memory items stored under the specified namespace. It returns a list of items that you can iterate through.

```python
items = store.search(namespace2)

for item in items:
    print(item.value)
```

### Semantic Search
Instead of just exact key lookups, we can configure the `InMemoryStore` with an embedding model to perform semantic search.
This allows the agent to retrieve memories that are conceptually similar to the user's query, which is crucial for handling large amounts of memory.

```python
from langchain_mistralai import MistralAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embedding_model = MistralAIEmbeddings(model='mistral-embed')
```

```python
store = InMemoryStore(index={'embed': embedding_model, 'dims':1536})
```

```python
namespace = ('users', 'u1')
```

```python
store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
# Semantic search allows querying the store with natural language.
# The 'limit' parameter restricts the number of returned memory items.
items = store.search(namespace, query="what is the user currently learning", limit=1)

for item in items:
    print(item.value)
```

```python
# Semantic search allows querying the store with natural language.
# The 'limit' parameter restricts the number of returned memory items.
items = store.search(namespace, query="what are user's preferences", limit=3)

for item in items:
    print(item.value)
```
