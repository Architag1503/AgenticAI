```python
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END
from langchain_mistralai import ChatMistralAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
class ParentState(TypedDict):

    question: str
    answer_eng: str
    answer_hin: str
    
```

```python
parent_llm = ChatMistralAI(model='mistral-small-2506')
subgraph_llm = ChatMistralAI(model='mistral-small-2506')
```

```python
def translate_text(state: ParentState):

    prompt = f"""
Translate the following text to Hindi.
Keep it natural and clear. Do not add extra content.

Text:
{state["answer_eng"]}
""".strip()
    
    translated_text = subgraph_llm.invoke(prompt).content

    return {'answer_hin': translated_text}
```

```python
# SUBGRAPH CONCEPT: We first define the subgraph just like any other StateGraph.
subgraph_builder = StateGraph(ParentState)

subgraph_builder.add_node('translate_text', translate_text)

subgraph_builder.add_edge(START, 'translate_text')
subgraph_builder.add_edge('translate_text', END)

# Compile the subgraph. A compiled graph acts as a Runnable, so it can be used as a node in another graph.
subgraph = subgraph_builder.compile()
```

```python
def generate_answer(state: ParentState):

    answer = parent_llm.invoke(f"You are a helpful assistant. Answer clearly.\n\nQuestion: {state['question']}").content
    return {'answer_eng': answer}
```

```python
parent_builder = StateGraph(ParentState)

parent_builder.add_node("answer", generate_answer)
# SUBGRAPH CONCEPT: We add the compiled subgraph as a node in the parent graph.
# When the parent graph reaches this node, it will execute the entire subgraph workflow.
parent_builder.add_node("translate", subgraph)

parent_builder.add_edge(START, 'answer')
parent_builder.add_edge('answer', 'translate')
parent_builder.add_edge('translate', END)
```

```python
graph = parent_builder.compile()

graph
```

```python
graph.invoke({'question': 'What is quantum physics'})
```
